## Worked example (adjacent data)
A small field: the take rises from 60% to 70%. Staying is still worth far more than the $20M exit. The indifference take shows how far the counterparty could push on economics alone.

```

def worked(d):
    p = {x.parameter: x.value for x in d['worked_example_field'].itertuples()}
    af = (1 - (1 + p['rate']) ** -p['years']) / p['rate']
    return {'annuity_factor': af, 'pv_before': p['profit_oil'] * (1 - p['take_before']) * p['annual_mbbl'] * af,
            'pv_after': p['profit_oil'] * (1 - p['take_after']) * p['annual_mbbl'] * af,
            'indifference_take': 1 - p['exit_value'] / (p['profit_oil'] * p['annual_mbbl'] * af)}

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
