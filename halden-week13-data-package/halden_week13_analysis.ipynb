## Worked example (adjacent data)
A worker who produces $50,000 of revenue at a $40,000 wage is worth hiring. A $100 raise at a 50% tax rate costs the firm $50.

```

def worked(d):
    p = {x.parameter: x.value for x in d['worked_example_labor'].itertuples()}
    return {'mrp': p['units_per_worker'] * p['price_per_unit'], 'mrp_minus_wage': p['units_per_worker'] * p['price_per_unit'] - p['wage'],
            'after_tax_wage_increase': p['wage_increase'] * (1 - p['tax_rate'])}

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
