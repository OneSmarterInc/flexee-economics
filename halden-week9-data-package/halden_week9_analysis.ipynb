## Worked example (adjacent data)
Two markets: A has strong local equity (rebranding loses 3c per fill), B has weak equity (gains 6c). Rebrand B, keep A.

```

def worked(d):
    p = {x.parameter: x.value for x in d['worked_example_markets'].itertuples()}
    gb = p['b_net_per_fill'] * p['fills_per_site_year'] * p['sites_each'] / 1e6
    cb = p['sites_each'] * p['cost_per_site'] / 1e6
    return {'market_b_gain_musd': gb, 'market_b_cost_musd': cb, 'market_b_payback_years': cb / gb,
            'market_a_gain_musd': p['a_net_per_fill'] * p['fills_per_site_year'] * p['sites_each'] / 1e6}

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
