## Worked example (adjacent data)
Entity A: EUR revenue 1,000, EUR costs 800. Net exposure is 200, not 800. Hedging the gross costs would create a loss four times larger than the real exposure.

```

def worked(d):
    p = {x.parameter: x.value for x in d['worked_example_entity'].itertuples()}
    chg = p['fx_post'] / p['fx_pre'] - 1
    return {'fx_change': chg, 'net_exposure': p['eur_revenue'] - p['eur_costs'],
            'net_impact': (p['eur_revenue'] - p['eur_costs']) * chg, 'wrong_hedge_on_gross_costs': p['eur_costs'] * chg}

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
