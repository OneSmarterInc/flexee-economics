## Worked example (adjacent data)
Station-level switching (elasticity −20) versus market-level demand (−0.03). Matching pays only when customers switch stations on small price gaps.

```

def worked(d):
    p = {x.parameter: x.value for x in d['worked_example_markets'].itertuples()}
    out = {}
    for k in ['elastic', 'inelastic']:
        at = p['station_elasticity_' + k] * (p['cut'] / p['pump'])
        out['ignore_cost_' + k] = p['margin'] * p['volume'] * abs(at)
    out['match_cost'] = p['cut'] * p['volume']
    return out

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
