The recession arrived in March. There is nothing new to decide; every decision you have already made is the decision. A recession is uneven: it hits each product by how far people can do without it, and each refinery by what it makes.

Unzip the data files next to this notebook before you run it.

```python
import pandas as pd

downturn = pd.read_csv('data/the_downturn.csv').set_index('parameter')['value']
products = pd.read_csv('data/how_products_fall.csv').set_index('product')['how_demand_follows_the_economy']
refineries = pd.read_csv('data/what_each_refinery_makes.csv').set_index('refinery')
due = pd.read_csv('data/what_comes_due.csv')
print(downturn)
```

## Part 1 — product by product

```python
g = downturn['economy_change']
hits = {p: products[p] * g for p in products.index}
for p, h in hits.items():
    print(f'{p:9s} demand {h * 100:+.2f}%')
```

## Part 2 — refinery by refinery

Each refinery's runs fall by its product mix, product by product.

```python
for key, r in refineries.iterrows():
    hit = sum(r[p] * hits[p] for p in ('gasoline', 'diesel', 'jet'))
    print(f"{r['label']:12s} runs {hit * 100:+.2f}%   (gasoline {r['gasoline']:.0%}, diesel {r['diesel']:.0%}, jet {r['jet']:.0%})")
print('The plant that makes the most jet fuel falls furthest.')
```

## Part 3 — Rotterdam against its shutdown point

```python
margin = downturn['european_margin_usd_bbl']
floor = downturn['rotterdam_shutdown_point_usd_bbl']
print(f'Europe margin ${margin:.2f} vs shutdown point ${floor:.2f}: ' + ('running still covers its own costs, barely' if margin > floor else 'pausing beats running'))
print(f"A pause costs ${downturn['rotterdam_restart_cost_musd']:.0f}M to undo when demand comes back.")
```
