Danielle wants the Halden name on all 4,300 US stations. Marcus says that in Louisiana and Mississippi the Cordell name is worth more. A brand shows up in one line of the accounts: what people buy in the shop. Work out what each name is worth per fill in each region, multiply by fills, and compare with the cost of paint.

Unzip the data files next to this notebook before you run it.

```python
import pandas as pd

regions = pd.read_csv('data/cordell_regions.csv')
cost = pd.read_csv('data/rebrand_cost.csv').set_index('parameter')['value']
margins = pd.read_csv('data/shop_margin_this_year.csv')
regions
```

## Part 1 — the worked example: two towns, one new sign

```python
ex = pd.read_csv('data/worked_example.csv').set_index('parameter')['value']
for town in ('a', 'b'):
    gain = ex[f'town_{town}_halden_minus_cordell_per_fill'] * ex['stations_each'] * ex['fills_per_station_per_year'] / 1e6
    paint = ex['stations_each'] * ex['cost_per_station_usd'] / 1e6
    print(f'Town {town.upper()}: gain ${gain:.2f}M a year against ${paint:.1f}M of paint' + (f', pays back in {paint / gain:.1f} years' if gain > 0 else ', never pays back'))
```

## Part 2 — the three regions, at the usual shop margin

```python
cost_per_station = cost['programme_cost_all_stations_musd'] * 1e6 / cost['stations']
r = regions.copy()
r['halden_minus_cordell'] = r['halden_would_add_per_fill_usd'] - r['cordell_adds_per_fill_usd']
r['gain_musd_year'] = r['halden_minus_cordell'] * r['stations'] * cost['fills_per_station_per_year'] / 1e6
r['paint_musd'] = r['stations'] * cost_per_station / 1e6
r['payback_years'] = r['paint_musd'] / r['gain_musd_year']
print(f'Cost per station: ${cost_per_station:,.0f}')
print(r[['label', 'stations', 'halden_minus_cordell', 'gain_musd_year', 'paint_musd', 'payback_years']].round(2))
print(f"All three regions: ${r['gain_musd_year'].sum():.2f}M a year. Only the regions where Halden earns more: ${r[r['gain_musd_year'] > 0]['gain_musd_year'].sum():.2f}M a year.")
```

## Part 3 — this year's shop margin changes the sums

What the whole class did about Pelican's price cut last summer set this year's shop margin. A brand's pull is scaled by it: this year's margin divided by the usual 42 cents.

```python
for m in margins.itertuples():
    scale = m.shop_margin_usd_gal / cost['usual_shop_margin_usd_gal']
    good = r[r['gain_musd_year'] > 0]
    gain = (good['gain_musd_year'] * scale).sum()
    print(f'{m.what_the_class_did_last_summer:12s} margin ${m.shop_margin_usd_gal:.3f}: the two paying regions earn ${gain:.2f}M a year, pay back in {good["paint_musd"].sum() / gain:.1f} years')
```
