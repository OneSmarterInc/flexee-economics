The Kessana government wants 74% of the field's profit oil instead of 62%. Halden's $2.8 billion is already in the ground. That money is irrelevant to the decision, and it is exactly why the government has the upper hand.

Unzip the data files next to this notebook before you run it.

```python
import pandas as pd

contract = pd.read_csv('data/the_contract.csv').set_index('parameter')['value']
field = pd.read_csv('data/the_field.csv').set_index('parameter')['value']
leaving = pd.read_csv('data/leaving_and_sunk.csv').set_index('parameter')['value']
takes = pd.read_csv('data/takes_to_test.csv').set_index('case')['government_take']
next_door = pd.read_csv('data/contracts_next_door.csv').set_index('contract')['government_take']
print(contract); print(field); print(leaving)
```

## Part 1 — a barrel of profit oil, and the field from here

Value of staying = profit oil × Halden's share × barrels a year × annuity factor. The annuity factor turns ten years of level cash into one number today.

```python
profit_oil = contract['long_run_brent_usd_bbl'] - contract['kessana_discount_usd_bbl'] - contract['lifting_cost_usd_bbl']
annual = field['reserves_left_million_bbl'] / field['years_of_production']
r, n = field['discount_rate'], int(field['years_of_production'])
annuity = (1 - (1 + r) ** -n) / r
print(f'profit oil ${profit_oil:.2f} a barrel; {annual:.0f} million barrels a year; annuity factor {annuity:.4f}')

def value_of_staying(take):
    return profit_oil * (1 - take) * annual * annuity

for case, take in takes.items():
    print(f'{case:9s} take {take:.0%}  Halden keeps ${profit_oil * (1 - take):6.2f} a barrel  staying is worth ${value_of_staying(take):8.0f}M')
print(f"leaving is worth ${leaving['exit_value_musd']:.0f}M")
```

## Part 2 — how far could they push?

```python
indifference = 1 - leaving['exit_value_musd'] / (profit_oil * annual * annuity)
print(f'Staying and leaving are worth the same at a take of {indifference:.1%}.')
print(f"Comparable contracts next door take between {next_door.min():.0%} and {next_door.max():.0%}; the demand is {contract['government_take_demanded']:.0%}.")
print('Notice that nothing above used the capital already spent.')
```
