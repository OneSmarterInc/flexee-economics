The Norwegian union wants 8% or the platforms stop. The Permian pays $145,000 and crews leave for a few thousand more. Baton Rouge's turnaround is due at the contractor peak. Three markets, built three ways.

Unzip the data files next to this notebook before you run it.

```python
import pandas as pd

union = pd.read_csv('data/the_union.csv').set_index('parameter')['value']
permian = pd.read_csv('data/the_permian.csv').set_index('parameter')['value']
turnaround = pd.read_csv('data/the_turnaround.csv').set_index('parameter')['value']
markets = pd.read_csv('data/three_markets.csv')
print(markets.to_string(index=False))
```

## Part 1 — Norway: the raise after tax

```python
gross = union['norwegian_wage_bill_musd_per_year'] * union['union_demand']
after_tax = gross * (1 - union['norway_tax_rate'])
print(f'The 8% costs ${gross:.1f}M a year before tax and ${after_tax:.2f}M after Norway\'s tax: {after_tax / gross:.0%} of the headline.')
```

## Part 2 — the Permian: what a worker is worth

```python
worth_k = permian['barrels_per_extra_worker_per_year'] * permian['margin_per_barrel_usd'] / 1000
print(f'One more Permian worker earns Halden ${worth_k:.1f}k a year against a ${permian["market_wage_k"]:.0f}k wage: {worth_k / permian["market_wage_k"]:.2f} times.')
print('Halden takes the market wage. The question is how to keep people, not what to pay them.')
```

## Part 3 — the turnaround: now or wait

```python
peak = turnaround['crews_off_peak_musd'] * turnaround['peak_multiple']
wait = turnaround['crews_off_peak_musd'] + turnaround['chance_of_breakdown_if_you_wait'] * turnaround['breakdown_cost_musd']
print(f'Now, at the peak: ${peak:.0f}M.  Wait: ${wait:.0f}M expected (crews plus the chance of a breakdown).')
print(f'Expected saving from waiting: ${peak - wait:.0f}M, {(peak - wait) / peak:.1%} of the peak cost, for {turnaround["plant_condition_penalty_points"]:.0f} points of plant condition.')
```
