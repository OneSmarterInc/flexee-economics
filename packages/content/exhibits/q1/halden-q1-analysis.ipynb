```

def worked(d):
    w = {x.parameter: x.value for x in d['worked_example_assets'].itertuples()}
    return {'x_reported_gross': w['x_revenue'] - w['x_cost'], 'x_posttax': (w['x_revenue'] - w['x_cost']) * (1 - w['x_tax']),
            'y_contribution': w['y_gross'] - w['y_variable'], 'y_net': w['y_gross'] - w['y_variable'] - w['y_fixed']}

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
