Estimate own-price elasticity for each market cluster from 156 weeks of price and volume.

The lesson: elasticity is not a constant you look up. It ranges from near zero where customers have nowhere to go, to strongly negative where they can switch.

The transmission problem: you set rack prices, dealers set street prices. A rack cut reaches the pump only partially.

```
import pandas as pd, numpy as np, math
d = {s: pd.read_csv(f'data/{s}.csv') for s in ['cluster_params', 'cluster_price_volume', 'pricing_params', 'fuel_nonfuel', 'worked_example_cluster', 'worked_example_params']}
for s, df in d.items(): print(s, df.shape)
```

## Worked example (adjacent data)
A sample cluster (design elasticity -0.08). Elasticity = SLOPE(ln volume, ln price). Response to a 5c rack cut = -elasticity x (cut x pass-through / pump) x 100.

```

def worked(d):
    w = d['worked_example_cluster']; p = {x.parameter: x.value for x in d['worked_example_params'].itertuples()}
    e = float(np.polyfit(w.ln_price, w.ln_volume, 1)[0])
    return {'worked_elasticity': e, 'worked_vol_response_pct': -e * (p['rack_cut'] * p['passthrough'] / p['pump_base']) * 100}

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
