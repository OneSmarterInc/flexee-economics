## Worked example (adjacent data)
Two projects under low and high carbon prices. Each wins in one scenario and loses in the other. That sign swing is why Week 12 is a bet, not an optimization.

```

def worked(d):
    p = {x.parameter: x.value for x in d['worked_example_projects'].itertuples()}
    return {'p1_low': p['p1_base'], 'p1_high': p['p1_base'] + p['p1_carbon_sens'] * (p['carbon_high'] - p['carbon_low']),
            'p2_low': p['p2_base'], 'p2_high': p['p2_base'] + p['p2_carbon_sens'] * (p['carbon_high'] - p['carbon_low'])}

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
