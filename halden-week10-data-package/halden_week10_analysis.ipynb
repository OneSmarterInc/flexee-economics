# Halden Energy — Week 10: The cycle turns — where the consequences converge
**Python path — loads the same canonical CSVs as the workbook.**

A recession arrives: GDP falls 3%. Demand falls unevenly by product — jet collapses, gasoline barely moves — so each refinery's product mix sets its exposure.

This is the convergence week. Five earlier decisions come due at once: Week 6 cancellable capex, Week 5 hedge coverage, the Week 4 transfer price (Delacroix's cover), Straits Pacific standing, and the Week 8 cash cushion.

Part A computes the demand hit. Part B applies the binding rules to a team's prior-week state and finds which constraints bind.

```
import pandas as pd, numpy as np, math
d = {s: pd.read_csv(f'data/{s}.csv') for s in ['product_elasticities', 'refinery_yields', 'recession_params', 'binding_rules', 'team_prior_state']}
for s, df in d.items(): print(s, df.shape)
```

## Worked example (adjacent data)
Worked on the disciplined reference team: compute Baton Rouge's product-mix hit, then apply the five binding rules. Nothing binds. That team has room to choose.

```

def worked(d):
    el = d['product_elasticities'].set_index('product')['income_elasticity']; g = d['recession_params']['value'].iloc[0]
    yl = d['refinery_yields'].set_index('refinery'); b = {x.parameter: x.value for x in d['binding_rules'].itertuples()}
    t = d['team_prior_state'].set_index('team').loc['reference_disciplined']
    out = {'br_hit': sum(yl.loc['Baton Rouge', p] * el[p] * g for p in ['gasoline', 'diesel', 'jet'])}
    out['disciplined_binding_count'] = int((t.cancellable_capex_musd < b['min_cancellable_capex_musd']) + (t.crude_hedge_coverage < b['min_crude_hedge_coverage'])
        + (t.br_reported_margin_strong == 1) + (t.straits_pacific_standing in ('strained', 'hostile')) + (t.cash_cushion_musd < b['min_cash_cushion_musd']))
    return out

wr = worked(d)
for k, v in wr.items(): print(f'{k:34s}', round(v, 4) if isinstance(v, (int, float)) else v)
```
